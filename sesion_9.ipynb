Introducción

Las máquinas logran tomar decisiones mediante algoritmos de aprendizaje automático, que analizan datos, encuentran patrones y utilizan esos patrones para clasificar o predecir resultados. Un ejemplo es MNIST, un conjunto de imágenes de números escritos a mano que se utiliza para entrenar modelos de inteligencia artificial.

MNIST – Clasificación binaria

En una clasificación binaria, la máquina debe elegir entre dos categorías. Por ejemplo, podemos utilizar MNIST para diferenciar entre los números 0 y 1.

El proceso sería:

Imagen → Datos/píxeles → Modelo → Decisión (0 o 1)

La máquina recibe los píxeles de la imagen como números y aprende cuáles características corresponden a cada número.
