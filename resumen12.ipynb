Introducción a conjuntos: Concepto de combinar varios modelos para obtener mejores predicciones.

Clasificadores por votación: Varios clasificadores hacen predicciones y se elige la clase mediante votación.

Bagging, Pasting y OOB: Técnicas que entrenan modelos con diferentes muestras de los datos; OOB permite evaluar usando datos que quedaron fuera del entrenamiento.

Bosques Aleatorios: Conjunto de árboles de decisión entrenados con muestras y características aleatorias para mejorar la generalización.

Algoritmos de Boosting: Construyen modelos de manera secuencial, haciendo que cada nuevo modelo se enfoque en corregir errores anteriores.

Stacking: Combina las predicciones de diferentes modelos mediante otro modelo, llamado meta-modelo, para producir la predicción final.

En resumen: el tema trata de cómo combinar varios modelos de machine learning para obtener predicciones más robustas y precisas que usando un solo modelo.
