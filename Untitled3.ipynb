CRISP-DM nació a finales de los años 90 como una metodología creada para establecer un proceso estándar para proyectos de minería de datos, aplicable a diferentes industrias.

Negocio: Se define qué problema tiene la empresa y qué se quiere lograr con el análisis de datos.
 Ejemplo: una empresa quiere saber por qué están disminuyendo sus ventas.
Datos: Se recopilan y se revisan los datos disponibles para conocer qué información sirve.
Ejemplo: ventas, clientes, productos, precios y fechas.
Preparación de datos: Se limpian y organizan los datos antes de analizarlos. Se eliminan errores, datos repetidos o información incompleta. Ejemplo: corregir precios incorrectos y eliminar clientes duplicados.
Modelado: Se utilizan técnicas o algoritmos para encontrar patrones y obtener resultados a partir de los datos preparados.
